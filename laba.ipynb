```python
from sqlalchemy import ForeignKey, create_engine, Column, Integer, String, DateTime,Date, select, text
from sqlalchemy.orm import DeclarativeBase,Mapped,mapped_column,relationship
from sqlalchemy.orm import sessionmaker

database_url = r'mssql+pyodbc://sas:1234@DESKTOP-PBMGJ7G\SENYAMSSQLSERVER/hico69_laba_sql?driver=ODBC+driver+17+for+SQL+Server'
engine = create_engine(database_url,echo=True)

class Base(DeclarativeBase):
    pass

class GROUPS(Base):
    __tablename__ = 'GROUPS'
    GROUP_ID : Mapped[int] = mapped_column(primary_key=True)
    NAME_GROUP: Mapped[str] = mapped_column(String(15),nullable=False)
    COURSE:Mapped[int]

class STUDENTS(Base):
    __tablename__ = 'STUDENTS'
    STUDENT_ID: Mapped[int] = mapped_column(primary_key=True)
    NAME_STUDENT: Mapped[str] = mapped_column(String(35),nullable=False)
    SURNAME_STUDENT: Mapped[str] = mapped_column(String(35),nullable=False)
    STUDENTS_SECONDNAME: Mapped[str] = mapped_column(String(35),nullable=False)
    BIRTHDAY: Mapped[DateTime] = mapped_column(Date)
    GROUP_ID: Mapped[int] = mapped_column(ForeignKey("GROUPS.GROUP_ID"))

class SUBJECTS(Base):
    __tablename__ = 'SUBJECTS'
    SUBJECT_ID: Mapped[int] = mapped_column(primary_key=True)
    SUBJECT_NAME: Mapped[str] = mapped_column(String(50),nullable=False)
    SUBJECT_HOURS: Mapped[int]

class PLANN(Base):
    __tablename__ = 'PLANN'
    SUBJECT_ID: Mapped[int] = mapped_column(ForeignKey("SUBJECTS.SUBJECT_ID"),primary_key=True)
    GROUP_ID: Mapped[int] = mapped_column(ForeignKey("GROUPS.GROUP_ID"),primary_key=True)

Base.metadata.create_all(engine)
```
